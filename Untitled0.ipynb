```python
# Устанавливаем библиотеки трансформеров и датасетов от Hugging Face
!pip install transformers datasets -q

import os
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score
from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer, DataCollatorWithPadding

train_df = pd.read_csv('/content/train (5).csv')
test_df = pd.read_csv('/content/test (4).csv')
print("Окружение Fine-Tuning готово, файлы на базе!")
```

```python
# 1. Создаем бланк под фолды
train_df['fold'] = -1

# 2. Настраиваем разбиение
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 3. # 3. Мы ТОЛЬКО генерируем индексы сплитов. Не нужно создавать датасеты или применять фильтры внутри этого цикла!

for fold, (train_idx, val_idx) in enumerate(skf.split(train_df, train_df['target'])):
    train_df.loc[val_idx, 'fold'] = fold

print("Creating a template for folds")

```

```python
# 1. Возвращаем проверенный и стабильный DistilBERT
model_name = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Жестко приводим target к инту в Pandas, зачищая любые косяки типов
train_df['target'] = train_df['target'].astype(int)

# 2. Переводим в формат Hugging Face
hf_train_dataset = Dataset.from_pandas(train_df[['text', 'target', 'fold']])
hf_test_dataset = Dataset.from_pandas(test_df[['text']])

# 3. Чистая функция токенизации
def tokenize_function(examples):
    tokenized = tokenizer(examples['text'], truncation=True)
    if 'target' in examples:
        tokenized['label'] = examples['target'] # Теперь тут гарантированно чистый int
    return tokenized

# 4. Токенизируем
tokenized_train_dataset = hf_train_dataset.map(tokenize_function, batched=True, remove_columns=['text', 'target'])
tokenized_test_dataset = hf_test_dataset.map(tokenize_function, batched=True, remove_columns=['text'])

print("Ячейка 3 перезапущена на DistilBERT!")

```

```python
# 1. Объявляем функцию подсчета F1-метрики
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)  # переводим логиты в 0 или 1
    f1 = f1_score(labels, predictions, average='binary')
    return {"f1": f1}

# 2. НАСТРОЙКА РУБИЛЬНИКА: Выбираем, какой фолд учим (сейчас Фолд 0)
current_fold = 0

# 3. Фильтруем данные из глобального tokenized_train_dataset (сделанного в Ячейке 3)
train_fold_data = tokenized_train_dataset.filter(lambda x: x['fold'] != current_fold)
val_fold_data = tokenized_train_dataset.filter(lambda x: x['fold'] == current_fold)

print(f"Данные для Фолда {current_fold} нарезаны!")
print(f"Размер трейна: {len(train_fold_data)}, Размер валидации: {len(val_fold_data)}")
```

```python
# 1. Сначала ПЕРЕИНИЦИАЛИЗИРУЕМ пустые бланки под предсказания всего датасета
oof_preds = np.zeros((len(train_df), 2))
test_preds = np.zeros((len(test_df), 2))

# 2. Инициализируем сборщик динамического паддинга батчей
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

# Запускаем цикл на 5 фолдов
for fold in range(5):
    print("\n" + "="*50)
    print(f"   СТАРТ ОБУЧЕНИЯ: ФОЛД {fold} из 5")
    print("="*50)

    # Шаг А: Нарезка данных лазером под текущий фолд
    train_fold_data = tokenized_train_dataset.filter(lambda x: x['fold'] != fold)
    val_fold_data = tokenized_train_dataset.filter(lambda x: x['fold'] == fold)

    # Шаг Б: Инициализация «свежей» копии модели (веса сбрасываются перед каждым фолдом!)
    model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)

    # Шаг В: Настройка пульта управления под текущий фолд
    training_args = TrainingArguments(
        output_dir=f"bert_fold_{fold}",
        learning_rate=2e-5,                     # Безопасный микро-шаг для DistilBERT
        per_device_train_batch_size=16,         # Оптимально для T4 GPU
        per_device_eval_batch_size=16,
        num_train_epochs=2,                     # 2 эпохи, чтобы не переобучить
        weight_decay=0.01,
        eval_strategy="epoch",
        save_strategy="epoch",
        load_best_model_at_end=True,            # Откатываемся на лучшую эпоху по F1
        metric_for_best_model="f1",
        fp16=True,                              # Турбо-режим GPU (ускорение х2)
        report_to="none"
    )

    # Шаг Г: Сборка трейнера
    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_fold_data,
        eval_dataset=val_fold_data,
        data_collator=data_collator,
        compute_metrics=compute_metrics
    )

    # Шаг Д: Запуск Fine-Tuning
    trainer.train()

    # Шаг Е: Забираем предсказания лучшей эпохи для ВАЛИДАЦИИ (в OOF копилку)
    val_predictions = trainer.predict(val_fold_data)
    val_idx = train_df[train_df['fold'] == fold].index
    oof_preds[val_idx] = val_predictions.predictions  # Записываем сырые логиты

    # Шаг Ж: Делаем предсказание для ТЕСТА текущей моделью и берем 1/5 её вклада
    test_predictions = trainer.predict(tokenized_test_dataset)
    test_preds += test_predictions.predictions / 5.0  # Усредняем (Блендинг фолдов)

    print(f"--- Обучение Фолда {fold} завершено! ---")

print("\n ОБУЧЕНИЕ ВСЕХ 5 ФОЛДОВ ПОЛНОСТЬЮ ЗАВЕРШЕНО!")
```

```python
# 1. Переводим финальные OOF-логиты всего датасета в классы 0 или 1
final_oof_predictions = np.argmax(oof_preds, axis=-1)

# 2. Считаем глобальный честный кросс-валидационный скор по всей таблице
final_f1_score = f1_score(train_df['target'], final_oof_predictions, average='binary')

print("="*50)
print(f"ИТОГОВЫЙ СТРАТИФИЦИРОВАННЫЙ OOF F1-SCORE ПАЙПЛАЙНА: {final_f1_score:.4f}")
print("="*50)

# 3. Формируем финальный submission.csv на основе усредненного мнения 5 моделей
final_test_predictions = np.argmax(test_preds, axis=-1)

submission = pd.DataFrame({
    'id': test_df['id'],
    'target': final_test_predictions
})

# Сохраняем итоговый файл
submission.to_csv('submission_distilbert_oof.csv', index=False)
print("Файл submission_distilbert_oof.csv успешно сгенерирован и готов к отправке на лидерборд!")




```
